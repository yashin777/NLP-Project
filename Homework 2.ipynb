# Step 5 Confidence Interval +    Hypothesis Test

Is average trestbps different between males and females?

```python
#female
trestbps0_mean=sample_200.loc[df_clean["sex"]==0,"trestbps"].mean()
trestbps0_mean
```

```python
trestbps0_standard_deviation=sample_200.loc[df_clean["sex"]==0,"trestbps"].std()
trestbps0_standard_deviation
```

```python
trestbps0_count=sample_200.loc[df_clean["sex"]==0,"trestbps"].count()
trestbps0_count
```

```python
#male
trestbps1_mean=sample_200.loc[df_clean["sex"]==1,"trestbps"].mean()
trestbps1_mean
```

```python
trestbps1_standard_deviation=sample_200.loc[df_clean["sex"]==1,"trestbps"].std()
trestbps1_standard_deviation
```

```python
trestbps1_count=sample_200.loc[df_clean["sex"]==1,"trestbps"].count()
trestbps1_count
```

```python
t_value0=2.0 #female
t_value1=1.98 #male
```

```python
import math
standard_error0=trestbps0_standard_deviation/math.sqrt(trestbps0_count)
standard_error0
```

```python
upper_bound0=trestbps0_mean+1.98*trestbps0_standard_deviation/math.sqrt(trestbps0_count)
upper_bound0
```

```python
lower_bound0=trestbps0_mean-1.98*trestbps0_standard_deviation/math.sqrt(trestbps0_count)
lower_bound0
```

```python
upper_bound1=trestbps1_mean+1.98*trestbps1_standard_deviation/math.sqrt(trestbps1_count)
upper_bound1
```

```python
lower_bound1=trestbps1_mean-1.98*trestbps1_standard_deviation/math.sqrt(trestbps1_count)
lower_bound1
```

```python
t_test=(trestbps0_mean-trestbps1_mean)/math.sqrt((trestbps0_standard_deviation**2/trestbps0_count)+(trestbps1_standard_deviation**2/trestbps1_count))
t_test
```

The interval inidicates the resting blood pressure intervals of female: [125.16 ,134.97 ] and male:[127.43,132.79] there is an overlapping region [127.43,132.79] because of this overlapping we cannot confidently say the means are different based on the confidence interval alone.<br>

The t-statistic value is -0.01687, alpha is 0.05 and the z value is 1.96 |-0.01687| < 1.96 which explains that we fail to reject the null hypotheis. We do not have evidence to say the average resting blood pressure is different between male and female.

```python
plt.hist(mean_list,bins=14,color='green')
plt.axvline(population_mean,color='orange',linestyle='dashed')
plt.xlabel('Sample Means of Trestbps')
plt.ylabel("Counts")
plt.title("Histogram of 200 Sample Means")
plt.show()
```

Yes, the distribution look normal because the histogram forms a bell-shaped curve. Sample variability means the difference in descriptive statistics such as the mean, between different samples sizes from the same population.For example, if different sample sizes from the population of 303 patient in this dataset, each sample might have a slight different average resting blood pressure. 

# Step 7 Ethical Reflection

  Bad sampling can lead to give unfair or incorrect results because it doesn't accurately represent the whole population. If the sample size is too small, it can lead to incorrect medical decisions, which is critical in healthcare. Bad sampling can result in biased conclusion which affects healthcare decisions.<br> 
  
 In this dataset, there are 303 patients. Samples that are closer to the population size (near 300) are more precise because they are better represented to the overall population. When the sample size is much smaller, the results become less precise.
