```python
import spacy
import pandas as pd
from tqdm import tqdm
import os

# ----------------------------
# Load dataset
# ----------------------------
df = pd.read_csv("metadata.csv")  # replace with your file path
df_clean = df["abstract"].dropna().str.lower()  # drop missing abstracts and lowercase
df_clean = df_clean.str.replace(r'[^a-z\s]', '', regex=True)  # remove punctuation/numbers

# ----------------------------
# Load spaCy model
# ----------------------------
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])

# ----------------------------
# Preprocessing function
# ----------------------------
def preprocess_text_chunked(texts, chunk_size=5000, batch_size=2000):
    """
    Preprocess text in chunks to avoid memory overload.
    - texts: list of strings
    - chunk_size: number of texts to process at once
    - batch_size: spaCy batch size
    """
    lemmatized_texts = []
    for start in tqdm(range(0, len(texts), chunk_size), desc="Processing chunks"):
        batch = texts[start:start+chunk_size]
        for doc in nlp.pipe(batch, batch_size=batch_size):
            words = [token.lemma_ for token in doc if token.is_alpha and not token.is_stop]
            lemmatized_texts.append(' '.join(words))
    return lemmatized_texts

# ----------------------------
# Process and save incrementally
# ----------------------------
output_dir = "processed_chunks"
os.makedirs(output_dir, exist_ok=True)

chunk_size_save = 50000  # save every 50k abstracts
for i, start in enumerate(range(0, len(df_clean), chunk_size_save)):
    batch = df_clean[start:start+chunk_size_save].tolist()
    processed = preprocess_text_chunked(batch)
    # Save each chunk to avoid memory issues
    pd.Series(processed).to_pickle(os.path.join(output_dir, f"clean_abstracts_{i}.pkl"))
    print(f"Saved chunk {i} ({len(processed)} abstracts)")

# ----------------------------
# Optional: merge all chunks into a final dataset
# ----------------------------
import glob

all_chunks = []
for f in sorted(glob.glob("processed_chunks/clean_abstracts_*.pkl")):
    all_chunks.extend(pd.read_pickle(f))

final_series = pd.Series(all_chunks)
final_series.to_pickle("clean_abstracts_final.pkl")
print("All chunks processed and final dataset saved!")
```
