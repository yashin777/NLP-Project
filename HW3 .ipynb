# Step 3 Baseline Model 1-k-Nearest Neighbors 

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report,confusion_matrix
X=df.drop("target",axis=1)
y=df["target"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
X_test_scaled
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled,y_train)
value_prediction=knn.predict(X_test_scaled)
print(classification_report(y_test,value_prediction))
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
accuracy1=accuracy_score(y_test,value_prediction)
precision1=precision_score(y_test,value_prediction)
recall1=recall_score(y_test,value_prediction)
f1_1=f1_score(y_test,value_prediction)
print("Accuracy:",accuracy1)
print("Precision:",precision1)
print("Recall:",recall1)
print("F1-score:",f1_score)
```

```python
confusionmatrix=confusion_matrix(y_test,value_prediction)
confusionmatrix
```

```python
# Threshold 0.2 
y_probabilties=knn.predict_proba(X_test_scaled)[:,1]
threshold=0.2
y_predict_threshold=(y_probabilties >= threshold).astype(int)
print(confusion_matrix(y_test,y_predict_threshold))
```

In this confusion matrix,  31 patients  were correctly predicted as having no heart diease, and 4 were false positve,predicted to have  heart diease when they actually did not. 36 patients  were correctly predicted as having heart diease, while 5 patients predicted as having no heart diease when they actually did. Although the count of 5 is small, it is  critical because these 5 patients do have heart diease. 

# Step 4  Baseline Model 2 -Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree_pipeline=DecisionTreeClassifier(criterion='gini',max_depth=3,random_state=42)
```

```python
tree_pipeline.fit(X_train,y_train)
```

```python
y_prediction=tree_pipeline.predict(X_test)
```

```python
print(classification_report(y_test,y_prediction))
```

```python
accuracy=accuracy_score(y_test,y_prediction)
precision=precision_score(y_test,y_prediction)
recall=recall_score(y_test,y_prediction)
f1=f1_score(y_test,y_prediction)
print("Accuracy:",accuracy)
print("Precision:",precision)
print("Recall:",recall)
print("F1-score:",f1)
 
```

```python
confusionmatrix1=confusion_matrix(y_test,y_prediction)
confusionmatrix1
```

Decision Trees do not require feature scaling because the algorithm learns by splitting the data into branches based on feature values.The algorithm use these branches to predict whether future patients have heart disease or not. <br>

In contrast, K-NN requires feature scaling because some features have maximum and minmum values that can affect the predictions, such as cholestrol, chest pain, thalach, and slope. After scaling, K-NN predicts by finding the 5 nearest data points and determining which label( 0 for no heart disease, 1 for heart disease) appears most frequently to make the prediction.

# Step 6 Reflection

For this project,  I learned the fundamentals of using K-Nearest Neighbors, and Decison Trees. K-NN works by selecting a specific number of nearby data points to make a prediction, while a Decison Tree splits the data into branches to make a prediction. K-NN is appropriate when dealing with  several features that contains maximum and minimum values, which need to be scaled before  comparing to the chosen number of data points.Scaling improves the precision for the prediction modeling.

Decison Tree are preferable for classification problems, such as  predicting whether a patient has heart disease in this dataset. It is also  recommended to apply cross validation to evaluate the model's accuracy and identify ways to improve performance metrics.  

The challenges in classification data preparation is ensuring that all categorical features, are correctly encoded and scaled when applying K-NN. For Decision Trees, using a higher max_depth can lead to  overfitting.
